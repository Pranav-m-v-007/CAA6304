```python

def detect_intrusion(protocol, port, data):

    if (protocol == "TCP" and port == 80
            and "SQL INJECTION" in data):
        print("Alert: SQL Injection Detected!")

    elif (protocol == "TCP" and port == 22
            and "FAILED LOGIN" in data):
        print("Alert: SSH Login Attack Detected!")

    else:
        print("Traffic is Normal")

# Test cases
detect_intrusion("TCP", 80, "SQL INJECTION detected")
detect_intrusion("TCP", 22, "FAILED LOGIN")
detect_intrusion("UDP", 53, "SQL INJECTION")
```

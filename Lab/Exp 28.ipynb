```python

from collections import Counter

logs = [
    "192.168.1.1",
    "192.168.1.2",
    "192.168.1.1",
    "192.168.1.1",
    "192.168.1.1",
    "192.168.1.3",
    "192.168.1.1",
    "192.168.1.1"
]

threshold = 4

request_count = Counter(logs)

print("IP Address Request Count")

for ip, count in request_count.items():
    print(ip, ":", count)

    if count > threshold:
        print("Alert: Possible DoS Attack from", ip)
```

```python

properties = [
    ("Encryption", "Confidentiality", True),
    ("Access Control", "Confidentiality", False),
    ("File Integrity", "Integrity", True),
    ("Data Validation", "Integrity", False),
    ("Backup System", "Availability", True),
    ("Server Uptime", "Availability", False)
]

violations = set()

for name, principle, status in properties:

    print(name, ":", status)

    if status == False:
        violations.add(principle)

print("\nCIA Violations:")

if violations:
    for item in sorted(violations):
        print(item, "is Violated")
else:
    print("No CIA violations detected")
```

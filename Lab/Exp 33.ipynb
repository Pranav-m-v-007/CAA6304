```python

action = input("Enter incident response action: ").lower()

if "training" in action or "planning" in action:
    print("Phase: Preparation")

elif "detect" in action or "investigate" in action:
    print("Phase: Identification")

elif "isolate" in action or "block" in action:
    print("Phase: Containment")

elif "remove" in action or "delete malware" in action:
    print("Phase: Eradication")

elif "restore" in action or "backup" in action:
    print("Phase: Recovery")

elif "review" in action or "improve" in action:
    print("Phase: Lessons Learned")

else:
    print("Phase: Unknown")
```

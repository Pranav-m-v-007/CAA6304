```python

logs = [
    "failed login from 192.168.1.10",
    "normal login from 192.168.1.20",
    "malicious IP 10.0.0.5 detected",
    "multiple failed login attempts",
    "unusual traffic detected"
]

print("HYPOTHESIS-BASED HUNTING")
for log in logs:
    if "failed login" in log:
        print("Suspicious:", log)

print("\nINTELLIGENCE-BASED HUNTING")
threat_ips = ["10.0.0.5", "192.168.1.99"]

for log in logs:
    for ip in threat_ips:
        if ip in log:
            print("Threat Intelligence Match:", log)

print("\nANALYTICS-BASED HUNTING")
for log in logs:
    if "unusual" in log or "multiple" in log:
        print("Anomaly Detected:", log)
```

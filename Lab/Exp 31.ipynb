```python

rules = [
    ("ALLOW", "TCP", 80),
    ("DENY", "TCP", 80),
    ("ALLOW", "UDP", 53)
]

def firewall(protocol, port):

    for action, rule_protocol, rule_port in rules:

        if protocol == rule_protocol and port == rule_port:
            return action

    return "DENY"

packets = [
    ("TCP", 80),
    ("UDP", 53),
    ("TCP", 22)
]

for protocol, port in packets:
    result = firewall(protocol, port)
    print(protocol, port, ":", result)
```

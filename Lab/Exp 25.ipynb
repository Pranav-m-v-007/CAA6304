```python

import hashlib
import json

records = [
    {"id": 1, "time": 100, "data": "Evidence collected"},
    {"id": 2, "time": 110, "data": "Evidence transferred"},
    {"id": 3, "time": 120, "data": "Evidence verified"}
]

previous_hash = "0"
valid = True

for record in records:
    content = json.dumps(record, sort_keys=True)
    current_hash = hashlib.sha256(
        (previous_hash + content).encode()
    ).hexdigest()

    print("Record:", record["id"])
    print("Hash:", current_hash[:16])

    previous_hash = current_hash

print("\nChain of custody verified successfully.")
```

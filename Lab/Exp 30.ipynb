```python

firewall_logs = ["192.168.1.10", "192.168.1.10"]
windows_logs = ["192.168.1.10", "192.168.1.10"]
vpn_logs = ["192.168.1.10"]

firewall_ip = set(firewall_logs)
windows_ip = set(windows_logs)
vpn_ip = set(vpn_logs)

print("SIEM LOG CORRELATION")

common_ips = firewall_ip & windows_ip & vpn_ip

for ip in common_ips:
    if (firewall_logs.count(ip) >= 2 and
        windows_logs.count(ip) >= 2):

        print("Suspicious IP:", ip)
        print("Alert: Possible Brute-Force Attack!")

if not common_ips:
    print("No correlated attack detected")
```

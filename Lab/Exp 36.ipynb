```python

description = input("Enter system description: ").lower()

if "credit card" in description or "payment" in description:
    print("Applicable Standard: PCI-DSS")

elif "patient" in description or "medical" in description:
    print("Applicable Standard: HIPAA")

elif "personal data" in description or "privacy" in description:
    print("Applicable Standard: GDPR")

elif "financial report" in description or "accounting" in description:
    print("Applicable Standard: SOX")

else:
    print("No matching standard found")
```

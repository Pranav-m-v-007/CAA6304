```python

import time

baseline = 3

login_events = [
    ("192.168.1.1", 1),
    ("192.168.1.2", 2),
    ("192.168.1.3", 5),
    ("192.168.1.4", 1)
]

print("Continuous Security Monitoring Started")

for ip, attempts in login_events:

    print("\nIP Address:", ip)
    print("Login Attempts:", attempts)

    if attempts > baseline:
        print("Alert: Abnormal Login Activity!")
    else:
        print("Normal Login Activity")

    time.sleep(1)

print("\nMonitoring Completed")
```

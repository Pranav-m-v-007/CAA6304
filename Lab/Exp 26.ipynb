```python

import hashlib

sample = "CreateFile VirtualAlloc WriteProcessMemory"
behavior_log = [
    "File created",
    "Process injection detected",
    "Network connection established"
]

# Static Analysis
file_hash = hashlib.sha256(sample.encode()).hexdigest()

print("STATIC ANALYSIS")
print("SHA256 Hash:", file_hash)

suspicious_apis = [
    "VirtualAlloc",
    "WriteProcessMemory",
    "CreateRemoteThread"
]

for api in suspicious_apis:
    if api in sample:
        print("Suspicious API Found:", api)

# Dynamic Analysis
print("\nDYNAMIC ANALYSIS")

for event in behavior_log:
    print("Behavior:", event)

    if "injection" in event.lower():
        print("Alert: Suspicious behavior detected!")
```

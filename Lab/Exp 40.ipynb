```python

task = input("Enter security task: ").lower()

if "network monitoring" in task or "packet analysis" in task:
    print("Recommended Tool: Wireshark")

elif "vulnerability scanning" in task:
    print("Recommended Tool: Nessus")

elif "penetration testing" in task:
    print("Recommended Tool: Metasploit")

elif "log monitoring" in task or "siem" in task:
    print("Recommended Tool: Splunk")

elif "web application testing" in task:
    print("Recommended Tool: Burp Suite")

elif "intrusion detection" in task:
    print("Recommended Tool: Snort")

elif "malware analysis" in task:
    print("Recommended Tool: Ghidra")

else:
    print("No suitable tool found")
```

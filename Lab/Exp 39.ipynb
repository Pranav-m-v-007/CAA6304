```python

# Raw IOC feed
ioc_feed = [
    "IP:192.168.1.100",
    "DOMAIN:malware.com",
    "IP:10.0.0.5"
]

# Internal security logs
internal_logs = [
    "Connection from 192.168.1.100",
    "Visited google.com",
    "Connection from 10.0.0.5"
]

# Normalize IOCs
normalized_iocs = []

for ioc in ioc_feed:
    indicator_type, value = ioc.split(":")
    normalized_iocs.append(value.lower().strip())

print("Normalized IOCs:")
for ioc in normalized_iocs:
    print(ioc)

# Correlation
print("\nThreat Intelligence Alerts:")

for log in internal_logs:
    for ioc in normalized_iocs:
        if ioc in log.lower():
            print("ALERT! IOC Match Found:", ioc)
            print("Log:", log)
```

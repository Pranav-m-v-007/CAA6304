```python

incident = input("Enter incident type: ").lower()

if "malware" in incident:
    print("Strategy: Isolate infected systems")

elif "phishing" in incident:
    print("Strategy: Block malicious email and URLs")

elif "dos" in incident:
    print("Strategy: Block suspicious IP addresses")

elif "data breach" in incident:
    print("Strategy: Restrict access and secure affected data")

elif "ransomware" in incident:
    print("Strategy: Disconnect affected systems and protect backups")

else:
    print("Strategy: Investigate and monitor the incident")
```
